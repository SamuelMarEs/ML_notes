## Excercise 1:

```python
college = pd.read_csv("college.csv")
```

```python
college
```

```python
college = college.rename({'Unnamed: 0': 'College'},axis=1)
college = college.set_index('College')
```

```python
college
```

```python
college.describe()
```

```python

pd.plotting.scatter_matrix(college[["Top10perc", "Apps", "Enroll"]])
```

```python
fig, ax = subplots(figsize=(8,8))
college.boxplot("Outstate", by="Private", ax=ax)
```

```python
# Creates a new variable "Elite" by BINNING the "Top10perc" variable into two groups based on whether or not the proportion 
# of students coming from the top 10% of their high school classes exceeds 50%
college['Elite'] = pd.cut(college['Top10perc'], [0,50,100], labels=['No', 'Yes'])
college
```

```python
college['Elite'].value_counts()
```

```python
fig, ax = subplots(figsize=(8,8))
college.boxplot("Outstate", by="Elite", ax=ax)
```

```python
fig, axes = subplots(ncols=2, nrows=2, figsize=(15,15))
college.plot.scatter("Apps", "Accept", ax=axes[0,0])
axes[0,0].set_title("Applications vs Accepance")
college.boxplot("Expend", by="Private", ax=axes[0,1])
college.plot.scatter("F.Undergrad", "P.Undergrad", ax=axes[1,1])
axes[1,1].set_title("Full-time vs Part-time undergraduates")
college.plot.scatter("PhD", "Grad.Rate", ax=axes[1,0])
axes[1,0].set_title("Faculty with PhD vs Graduation rate")
```

```python
Auto
```

```python
Auto = pd.read_csv('Auto.csv', na_values=['?'], index_col="name")
```

```python
Auto
```

```python
for column in Auto:
    print(f'Range of {column}: ({Auto[column].min()}, {Auto[column].max()})')
```

```python
for column in Auto:
    print(f'Mean and standard deviation of {column}: mean = {Auto[column].mean()}, std = {Auto[column].std()}')
```

```python
Auto2 = Auto.drop(Auto.index[10:85])
```

```python
Auto2
```

```python
for column in Auto2:
    print(f'Mean and standard deviation of {column}: mean = {Auto2[column].mean()}, std = {Auto2[column].std()}')
```

```python
pd.plotting.scatter_matrix(Auto2, figsize=(10,10))
```

```python
for column in Auto.columns:
    if column != 'mpg':
        if column not in ["cylinders", "year", "origin"]:
            Auto.plot.scatter(x=column, y='mpg', figsize=(8,8), title=f"MPG vs {column}")
        else:
            Auto.boxplot('mpg', by=column, figsize=(8,8))
```

The plots suggest that there's a correlation between "mpg", and all three of horse power, wight and displacement. The number of miles per gallon seems to decrease as this three variables increase. As for the years, there appears to be an increase in mpg as the years go on, but the miles per gallon seem to decrease as the number of cylinders increases.
