The np.random.normal() function generates a vector of random normal distributed variables. Arguments are loc (mean), scale (std), and size (number of elements). By default it generates a single random variable with loc=0 and scale=1.

```python
x = np.random.normal(size=50)
x
```

```python
y = x + np.random.normal(loc=50, scale=1, size=50)
np.corrcoef(x,y)        # Computes the correlation matrix between x and y.
```

We can set a *random seed* using the np.random.default_rng() function. This function takes an arbitrary integer. If we set a random seed before generating random data, then the re-running our code wil yield the same results. The object rng has all the random generating methods found in np.random, thus we can generate normal data in the following way:

```python
rng = np.random.default_rng(1303)
print(rng.normal(scale=5, size=2))
rng2 = np.random.default_rng(1303)
print(rng2.normal(scale=5, size=2))
# This also allows for getting the same result as in the textbook wen generating random data.
```

```python
rng = np.random.default_rng(3)
y = rng.standard_normal(10)
np.mean(y), y.mean()            # Calculates the mean of a set of data
```

```python
np.var(y), y.var(), np.mean((y-y.mean())**2)        # Calculates the variance of a set of data
```

```python
np.sqrt(np.var(y)), np.std(y)       # Calculates standard deviation of a set of data
```

```python
X = rng.standard_normal((10,3))     # Creates a 10x3 matrix of N(0,1) random variables
X
```

```python
X.mean(axis=0), X.mean(0)              # Computes the rows mean
```

# Indexing Data

```python
A = np.array(np.arange(16)).reshape((4,4))
A
```

```python
A[1,2]          # Item at row 2, column 3
```

```python
A[[1,3]]        # Rows 2 and 4
```

```python
A[:,[0,2]]      # All the rows, and columns 1
```

```python
A[[1,3],[0,2]]      # Provides pairs of i,j indices for a series of entries, i.e. the elements at [1,3] and [0,2]
```

```python
A[[1,3]][:,[0,2]]       # Extracts the submatrix of rows 2 and 3, and columns 1 and 3
```

```python
idx = np.ix_([1,3],[0,2,3])
A[idx]
```

```python
# Boolean is a type equivalent to True or False (1 or 0)
keep_rows = np.zeros(A.shape[0], bool)
keep_rows
```

```python
keep_rows[[1,3]] = True         # We set the second and fourth elements of the list to true
keep_rows
```

```python
np.all(keep_rows == np.array([0,1,0,1]))
```

```python
A[np.array([0,1,0,1])]          # first, second, first, second rows of A
```

```python
A[keep_rows]                    # second and fourth rows of A (rows for which the Boolean equals True)
```

```python
keep_cols = np.zeros(A.shape[1], bool)
keep_cols[[0,2,3]] = True
idx_bool = np.ix_(keep_rows, keep_cols)
A[idx_bool]
```

Suppose that we want to create a data frame consisting of the *weight* and *origin* of the subset of cars with *year* greater tham 80, i.e. those built after 1980.

```python
# We create a Boolean to select only those rows whose year is higher than 80 for the selected columns
idx_80 = Auto_re['year'] > 80
Auto_re.loc[idx_80, ['weight', 'origin']]
```

```python
# This can be also done more concisely, using a lambda function
Auto_re.loc[lambda df: df['year'] > 80, ['weight', 'origin']]
# This lambda creates a function that takes a single argument, here df, and returns df['year']>80
```

```python
# For another example, assume we want cars built after 1980 that achieve geater than 30 miles per gallon
Auto_re.loc[lambda df: (df['year']>80) & (df['mpg']>30),
             ['weight', 'origin']
             ]
```

```python
# Now assume we want ford and datsun cars with displacement less than 300
Auto_re.loc[lambda df: (df['displacement']<300) & (df.index.str.contains('ford') | df.index.str.contains('datsun')),
            ['weight', 'origin']
            ]
```

# String Formatting

```python
rng = np.random.default_rng(1)
A = rng.standard_normal((127,5))
M = rng.choice([0, np.nan], p=[0.8, 0.2], size=A.shape)     # rng.choice() is used to overwrite some random entries
A += M
D = pd.DataFrame(A, columns=['food',
                             'bar',
                             'pickle',
                             'snack',
                             'popcorn'])
D[:3]
```

```python
for col in D.columns:
    template = 'Column "{0}" has {1:.2%} missing values'        # {1:.2%} indicates that the second argument is a percent with two decimal digits
    print(template.format(col, np.isnan(D[col]).mean()))
```
